```python
# Task 2
import numpy as np

A = np.array([
    [2, 1, 3],
    [1, 2, 1],
    [3, 1, 2]
])

B = np.array([
    [1, 2, 1],
    [2, 1, 3],
    [1, 3, 2]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

addition = A + B
print("\nMatrix Addition:")
print(addition)

elementwise_multiplication = A * B
print("\nElement-wise Multiplication:")
print(elementwise_multiplication)

matrix_multiplication = A @ B
print("\nMatrix Multiplication:")
print(matrix_multiplication)

transpose_A = A.T
print("\nTranspose of A:")
print(transpose_A)

det_A = np.linalg.det(A)
print("\nDeterminant of A:", det_A)


if not np.isclose(det_A, 0):
    inverse_A = np.linalg.inv(A)
    print("\nInverse of A:")
    print(inverse_A)

verification = A @ inverse_A
print("\nA @ A^-1:")
print(verification)

print("\nIs A @ A^-1 equal to identity matrix?")
print(np.allclose(verification, np.eye(3)))
```

```python
import matplotlib.pyplot as plt

np.random.seed(42)

mean = 50
std_dev = 10

data = np.random.normal(
    loc=mean,
    scale=std_dev,
    size=1000
)

print("First 10 values:")
print(data[:10])

sample_mean = np.mean(data)
print("\nChosen Mean:", mean)
print("Sample Mean:", sample_mean)

sample_std = np.std(data, ddof=1)
print("Chosen Standard Deviation:", std_dev)
print("Sample Standard Deviation:", sample_std)

mean_difference = sample_mean - mean
std_difference = sample_std - std_dev
print("\nComparison:")
print("Difference in Mean:", mean_difference)
print("Difference in Standard Deviation:", std_difference)

plt.hist(data, bins=30, edgecolor="black")
plt.title("Histogram of Normally Distributed Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()

```

```python
df = pd.read_csv("train.csv")
print(df.head(10))
print(df.tail(10))
print(df.shape)
print(df.columns)
print(df.info())
print(df.describe())

print(df.loc[0])
print(df.loc[0, "Name"])

print(df.iloc[0])
print(df.iloc[0, 3])
```

```python
# Task 5

male_over_50 = df[
    (df["Sex"] == "male") &
    (df["Age"] > 50)
]

print("\n1. Male passengers older than 50:")
print("Count:", len(male_over_50))



female_first_class = df[
    (df["Sex"] == "female") &
    (df["Pclass"] == 1)
]
female_first_class_count = len(female_first_class)

female_first_class_survival_rate = (
    female_first_class["Survived"].mean() * 100
)

print("\n2. Female first-class passengers:")
print("Count:", female_first_class_count)
print("Percentage survived:",
      female_first_class_survival_rate, "%")
overall_median_fare = df["Fare"].median()



age_20_40_high_fare_survived = df[
    (df["Age"] >= 20) &
    (df["Age"] <= 40) &
    (df["Fare"] > overall_median_fare) &
    (df["Survived"] == 1)
]
print("\n3. Age 20-40, Fare above overall median, and survived:")
print("Overall median Fare:", overall_median_fare)
print("Count:", len(age_20_40_high_fare_survived))


travelling_alone_under_30_did_not_survive = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

print("\n4. Travelling alone, age < 30, and did not survive:")
print("Count:", len(travelling_alone_under_30_did_not_survive))



southampton_data = df[df["Embarked"] == "S"]
southampton_median_fare = southampton_data["Fare"].median()
southampton_class_2_3_high_fare = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median_fare)
]
print("\n5. Embarked = S, Pclass 2 or 3,")
print("   Fare above Southampton median:")
print("Southampton median Fare:", southampton_median_fare)
print("Count:", len(southampton_class_2_3_high_fare))
```

```python
# Task 6
survival_rate_by_sex = (
    df.groupby("Sex")["Survived"]
    .mean()
    * 100
)

print("\n1. Survival rate by Sex:")
print(survival_rate_by_sex)

survival_rate_by_pclass = (
    df.groupby("Pclass")["Survived"]
    .mean()
    * 100
)
print("\n2. Survival rate by Pclass:")
print(survival_rate_by_pclass)


average_age_fare_by_pclass = (
    df.groupby("Pclass")[["Age", "Fare"]]
    .mean()
)
print("\n3. Average Age and Fare by Pclass:")
print(average_age_fare_by_pclass)


sex_pclass_summary = (
    df.groupby(["Sex", "Pclass"])
    .agg(
        Passenger_Count=("PassengerId", "count"),
        Survival_Rate=("Survived", "mean")
    )
)

sex_pclass_summary["Survival_Rate"] = (
    sex_pclass_summary["Survival_Rate"] * 100
)

print("\n4. Passenger count and survival rate by Sex-Pclass:")
print(sex_pclass_summary)


embarked_summary = (
    df.groupby("Embarked")
    .agg(
        Passenger_Count=("PassengerId", "count"),
        Average_Fare=("Fare", "mean"),
        Survival_Rate=("Survived", "mean")
    )
)

embarked_summary["Survival_Rate"] = (
    embarked_summary["Survival_Rate"] * 100
)

print("\n5. Passenger count, average Fare, and survival rate by Embarked:")
print(embarked_summary)
```

```python
missing_count = df.isnull().sum()

missing_percentage = (
    df.isnull().sum() / len(df)
) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

print("\nMissing values summary:")
print(missing_summary)

plt.figure(figsize=(10, 5))

missing_count.plot(kind="bar")

plt.title("Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

print("\nMissing Age values BEFORE filling:")
print(df["Age"].isnull().sum())

age_mean = df["Age"].mean()
df["Age"] = df["Age"].fillna(age_mean)
print("Missing Age values AFTER filling:")
print(df["Age"].isnull().sum())


age_mean_imputed = df["Age"].copy()
age_median_imputed = df["Age"].copy()
age_mode_imputed = df["Age"].copy()
age_random_imputed = df["Age"].copy()
# Mean imputation

age_mean_imputed = age_mean_imputed.fillna(
    age_mean_imputed.mean()
)

# Median imputation
age_median_imputed = age_median_imputed.fillna(
    age_median_imputed.median()
)

# Mode imputation
age_mode = age_mode_imputed.mode()[0]
age_mode_imputed = age_mode_imputed.fillna(
    age_mode
)
# Random value imputation

np.random.seed(42)

missing_indices = age_random_imputed[
    age_random_imputed.isnull()
].index

observed_age_values = age_random_imputed.dropna().values

random_values = np.random.choice(
    observed_age_values,
    size=len(missing_indices)
)

age_random_imputed.loc[missing_indices] = random_values


print("\nImputation methods completed:")
print("Mean imputation")
print("Median imputation")
print("Mode imputation")
print("Random value imputation")
```
