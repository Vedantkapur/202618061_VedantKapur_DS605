```python
# ==========================
# Task 2- Vectorized Arithmetic and Linear Algebra
# ==========================
import numpy as np

A = np.array([
    [2, 1, 3],
    [1, 2, 1],
    [3, 1, 2]
])

B = np.array([
    [1, 2, 1],
    [2, 1, 3],
    [1, 3, 2]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

addition = A + B
print("\nMatrix Addition:")
print(addition)

elementwise_multiplication = A * B
print("\nElement-wise Multiplication:")
print(elementwise_multiplication)

matrix_multiplication = A @ B
print("\nMatrix Multiplication:")
print(matrix_multiplication)

transpose_A = A.T
print("\nTranspose of A:")
print(transpose_A)

det_A = np.linalg.det(A)
print("\nDeterminant of A:", det_A)


if not np.isclose(det_A, 0):
    inverse_A = np.linalg.inv(A)
    print("\nInverse of A:")
    print(inverse_A)

verification = A @ inverse_A
print("\nA @ A^-1:")
print(verification)

print("\nIs A @ A^-1 equal to identity matrix?")
print(np.allclose(verification, np.eye(3)))
```

```python
import matplotlib.pyplot as plt
# =============================
# Task 3 - Normal Distribution and Histogram
# =============================

np.random.seed(42)

mean = 50
std_dev = 10

data = np.random.normal(
    loc=mean,
    scale=std_dev,
    size=1000
)

print("First 10 values:")
print(data[:10])

sample_mean = np.mean(data)
print("\nChosen Mean:", mean)
print("Sample Mean:", sample_mean)

sample_std = np.std(data, ddof=1)
print("Chosen Standard Deviation:", std_dev)
print("Sample Standard Deviation:", sample_std)

mean_difference = sample_mean - mean
std_difference = sample_std - std_dev
print("\nComparison:")
print("Difference in Mean:", mean_difference)
print("Difference in Standard Deviation:", std_difference)

plt.hist(data, bins=30, edgecolor="black")
plt.title("Histogram of Normally Distributed Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.savefig("Figures/missing_values.png")
plt.show()

```

```python
# ===========================================
# Task 4 - Load and Inspect Data
# ===========================================

df = pd.read_csv("train.csv")
print(df.head(10))
print(df.tail(10))
print(df.shape)
print(df.columns)
print(df.info())
print(df.describe())

print(df.loc[0])
print(df.loc[0, "Name"])

print(df.iloc[0])
print(df.iloc[0, 3])
```

```python
# ===========================================
# Task 5 Filtering and Querying
# ===========================================

male_over_50 = df[
    (df["Sex"] == "male") &
    (df["Age"] > 50)
]

print("\n1. Male passengers older than 50:")
print("Count:", len(male_over_50))



female_first_class = df[
    (df["Sex"] == "female") &
    (df["Pclass"] == 1)
]
female_first_class_count = len(female_first_class)

female_first_class_survival_rate = (
    female_first_class["Survived"].mean() * 100
)

print("\n2. Female first-class passengers:")
print("Count:", female_first_class_count)
print("Percentage survived:",
      female_first_class_survival_rate, "%")
overall_median_fare = df["Fare"].median()



age_20_40_high_fare_survived = df[
    (df["Age"] >= 20) &
    (df["Age"] <= 40) &
    (df["Fare"] > overall_median_fare) &
    (df["Survived"] == 1)
]
print("\n3. Age 20-40, Fare above overall median, and survived:")
print("Overall median Fare:", overall_median_fare)
print("Count:", len(age_20_40_high_fare_survived))


travelling_alone_under_30_did_not_survive = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

print("\n4. Travelling alone, age < 30, and did not survive:")
print("Count:", len(travelling_alone_under_30_did_not_survive))



southampton_data = df[df["Embarked"] == "S"]
southampton_median_fare = southampton_data["Fare"].median()
southampton_class_2_3_high_fare = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median_fare)
]
print("\n5. Embarked = S, Pclass 2 or 3,")
print("   Fare above Southampton median:")
print("Southampton median Fare:", southampton_median_fare)
print("Count:", len(southampton_class_2_3_high_fare))
```

```python
# ===========================================
# Task 6
# ===========================================

survival_rate_by_sex = (
    df.groupby("Sex")["Survived"]
    .mean()
    * 100
)

print("\n1. Survival rate by Sex:")
print(survival_rate_by_sex)

survival_rate_by_pclass = (
    df.groupby("Pclass")["Survived"]
    .mean()
    * 100
)
print("\n2. Survival rate by Pclass:")
print(survival_rate_by_pclass)


average_age_fare_by_pclass = (
    df.groupby("Pclass")[["Age", "Fare"]]
    .mean()
)
print("\n3. Average Age and Fare by Pclass:")
print(average_age_fare_by_pclass)


sex_pclass_summary = (
    df.groupby(["Sex", "Pclass"])
    .agg(
        Passenger_Count=("PassengerId", "count"),
        Survival_Rate=("Survived", "mean")
    )
)

sex_pclass_summary["Survival_Rate"] = (
    sex_pclass_summary["Survival_Rate"] * 100
)

print("\n4. Passenger count and survival rate by Sex-Pclass:")
print(sex_pclass_summary)


embarked_summary = (
    df.groupby("Embarked")
    .agg(
        Passenger_Count=("PassengerId", "count"),
        Average_Fare=("Fare", "mean"),
        Survival_Rate=("Survived", "mean")
    )
)

embarked_summary["Survival_Rate"] = (
    embarked_summary["Survival_Rate"] * 100
)

print("\n5. Passenger count, average Fare, and survival rate by Embarked:")
print(embarked_summary)
```

```python
# ============================================
# Task 7
# ============================================

missing_count = df.isnull().sum()

missing_percentage = (
    df.isnull().sum() / len(df)
) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

print("\nMissing values summary:")
print(missing_summary)

plt.figure(figsize=(10, 5))

missing_count.plot(kind="bar")

plt.title("Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig("Figures/missing_values1.png")
plt.show()

print("\nMissing Age values BEFORE filling:")
print(df["Age"].isnull().sum())

age_mean = df["Age"].mean()
df["Age"] = df["Age"].fillna(age_mean)
print("Missing Age values AFTER filling:")
print(df["Age"].isnull().sum())


age_mean_imputed = df["Age"].copy()
age_median_imputed = df["Age"].copy()
age_mode_imputed = df["Age"].copy()
age_random_imputed = df["Age"].copy()
# Mean imputation

age_mean_imputed = age_mean_imputed.fillna(
    age_mean_imputed.mean()
)

# Median imputation
age_median_imputed = age_median_imputed.fillna(
    age_median_imputed.median()
)

# Mode imputation
age_mode = age_mode_imputed.mode()[0]
age_mode_imputed = age_mode_imputed.fillna(
    age_mode
)
# Random value imputation

np.random.seed(42)

missing_indices = age_random_imputed[
    age_random_imputed.isnull()
].index

observed_age_values = age_random_imputed.dropna().values

random_values = np.random.choice(
    observed_age_values,
    size=len(missing_indices)
)

age_random_imputed.loc[missing_indices] = random_values


print("\nImputation methods completed:")
print("Mean imputation")
print("Median imputation")
print("Mode imputation")
print("Random value imputation")

Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

fare_outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]

print("\nFare Outlier Analysis:")

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print("Number of Fare Outliers:", len(fare_outliers))
```

```python
#============================================================
# TASK 8 - FEATURES AND PIVOT TABLE
#============================================================

print("\n")
print("=" * 60)
print("TASK 8 - FEATURES AND PIVOT TABLE")
print("=" * 60)
# Creating familysize
df["FamilySize"] = (
    df["SibSp"] +
    df["Parch"] +
    1
)

print("FamilySize created:")
print(df[["SibSp", "Parch", "FamilySize"]].head())

df["IsAlone"] = (
    df["FamilySize"] == 1
).astype(int)

print("\nIsAlone created:")
print(df[["FamilySize", "IsAlone"]].head())


pivot_table = pd.pivot_table(
    df,
    index="Sex",
    columns="Pclass",
    values="Survived",
    aggfunc="mean"
)

print("\nSurvival Rate Pivot Table:")
print(pivot_table)


highest_group = pivot_table.stack().idxmax()
highest_rate = pivot_table.stack().max()

lowest_group = pivot_table.stack().idxmin()
lowest_rate = pivot_table.stack().min()

print("Highest survival group:")
print("Sex:", highest_group[0])
print("Pclass:", highest_group[1])
print("Survival Rate:", highest_rate * 100, "%")


print("Lowest survival group:")
print("Sex:", lowest_group[0])
print("Pclass:", lowest_group[1])
print("Survival Rate:", lowest_rate * 100, "%")


```

```python
import seaborn as sns

# ============================================================
# TASK 9 - VISUALIZATIONS AND OBSERVATIONS
# ============================================================

numerical_columns = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "FamilySize",
    "IsAlone"
]

correlation_matrix = df[numerical_columns].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.savefig("Figures/missing_values2.png")
plt.show()

corr_pairs = correlation_matrix.where(
    np.triu(
        np.ones(correlation_matrix.shape),
        k=1
    ).astype(bool)
).stack()

strongest_positive = corr_pairs.idxmax()
strongest_positive_value = corr_pairs.max()

strongest_negative = corr_pairs.idxmin()
strongest_negative_value = corr_pairs.min()

print("\nStrongest positive relationship:")
print(
    strongest_positive,
    "=",
    strongest_positive_value
)

print("\nStrongest negative relationship:")
print(
    strongest_negative,
    "=",
    strongest_negative_value
)

survival_by_sex = (
    df.groupby("Sex")["Survived"]
    .mean()
    * 100
)

print("\nSurvival Rate by Sex:")
print(survival_by_sex)


plt.figure(figsize=(7, 5))

survival_by_sex.plot(
    kind="bar"
)

plt.title("Survival Rate by Sex")
plt.xlabel("Sex")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.savefig("Figures/missing_values3.png")
plt.show()


plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Survived"
)

plt.title("Age vs Fare by Survival")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.tight_layout()
plt.savefig("Figures/missing_values4.png")
plt.show()


print("=" * 60)
print("OBSERVATIONS")
print("=" * 60)


male_survival = survival_by_sex["male"]
female_survival = survival_by_sex["female"]

print(
    f"1. Female passengers had a survival rate of "
    f"{female_survival:.2f}%, while male passengers had "
    f"a survival rate of {male_survival:.2f}%."
)

print(
    f"2. The strongest positive correlation was between "
    f"{strongest_positive[0]} and {strongest_positive[1]}, "
    f"with a correlation of {strongest_positive_value:.2f}."
)

print(
    f"3. The strongest negative correlation was between "
    f"{strongest_negative[0]} and {strongest_negative[1]}, "
    f"with a correlation of {strongest_negative_value:.2f}."
)

print(
    f"4. The median Fare was {df['Fare'].median():.2f}, "
    f"and {len(fare_outliers)} passengers were identified "
    f"as Fare outliers using the 1.5 × IQR rule."
)

print(
    f"5. The average passenger age was "
    f"{df['Age'].mean():.2f} years after filling missing "
    f"Age values with the mean."
)

print(
    f"6. The highest survival group in the Sex-Pclass "
    f"pivot table was {highest_group[0]} passengers in "
    f"Pclass {highest_group[1]}, with a survival rate of "
    f"{highest_rate * 100:.2f}%."
)

print(
    f"7. The lowest survival group was {lowest_group[0]} "
    f"passengers in Pclass {lowest_group[1]}, with a "
    f"survival rate of {lowest_rate * 100:.2f}%."
)
```

```python
# ============================================================
# FINAL DATA CLEANING
# ============================================================

df["Age"] = df["Age"].fillna(df["Age"].mean())

df["Embarked"] = df["Embarked"].fillna(
    df["Embarked"].mode()[0]
)

df["Cabin"] = df["Cabin"].fillna("Unknown") #Filling the values that were empty with unknown


# Save cleaned dataset
df.to_csv(
    "cleaned_titanic.csv",
    index=False
)


print("Cleaned Titanic dataset saved successfully.")

print("Missing values after cleaning:")
print(df.isnull().sum())
```
